```python
import numpy as np


GRID_SIZE = 5
START_STATE = (0, 0)
GOAL_STATE = (GRID_SIZE - 1, GRID_SIZE - 1)
ACTIONS = ['up', 'down', 'left', 'right']
NUM_ACTIONS = len(ACTIONS)


LEARNING_RATE = 0.8
DISCOUNT_FACTOR = 0.95
EXPLORATION_RATE = 0.1
NUM_EPISODES = 1000


q_table = np.zeros((GRID_SIZE, GRID_SIZE, NUM_ACTIONS))

def get_reward(state, next_state):
    if next_state == GOAL_STATE:
        return 100
    elif state == next_state:
        return -10
    else:
        return -1


def get_next_state(state, action):
    row, col = state
    if action == 'up':
        new_row = max(0, row - 1)
        return (new_row, col)
    elif action == 'down':
        new_row = min(GRID_SIZE - 1, row + 1)
        return (new_row, col)
    elif action == 'left':
        new_col = max(0, col - 1)
        return (row, new_col)
    elif action == 'right':
        new_col = min(GRID_SIZE - 1, col + 1)
        return (row, new_col)
    return state

def choose_action(state):
    if np.random.uniform(0, 1) < EXPLORATION_RATE:
        return np.random.choice(ACTIONS)
    else:
        row, col = state
        return ACTIONS[np.argmax(q_table[row, col, :])]

for episode in range(NUM_EPISODES):
    current_state = START_STATE
    done = False

    while not done:
        action = choose_action(current_state)
        action_index = ACTIONS.index(action)
        next_state = get_next_state(current_state, action)
        reward = get_reward(current_state, next_state)

        old_q_value = q_table[current_state[0], current_state[1], action_index]
        max_next_q = np.max(q_table[next_state[0], next_state[1], :])
        new_q_value = old_q_value + LEARNING_RATE * (reward + DISCOUNT_FACTOR * max_next_q - old_q_value)
        q_table[current_state[0], current_state[1], action_index] = new_q_value

        current_state = next_state

        if current_state == GOAL_STATE:
            done = True


optimal_policy = np.empty((GRID_SIZE, GRID_SIZE), dtype='U10')
for r in range(GRID_SIZE):
    for c in range(GRID_SIZE):
        optimal_policy[r, c] = ACTIONS[np.argmax(q_table[r, c, :])]

print("Q-table after training:")
print(q_table)
print("\nOptimal Policy:")
print(optimal_policy)
```
