```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(0, 10.1, 0.1)
```

```python
def triangular(x, a, b, c):
    return np.maximum(np.minimum((x - a) / (b - a), (c - x) / (c - b)), 0)
```

```python
def trapezoidal(x, a, b, c, d):
    return np.maximum(np.minimum(np.minimum((x - a) / (b - a), 1), (d - x) / (d - c)),0)
```

```python
def bell(x, a, b, c):
    return 1 / (1 + np.abs((x - c) / a) ** (2 * b))
```

```python
tri = triangular(x, 2, 5, 8)
trap = trapezoidal(x, 2, 4, 6, 8)
bell_curve = bell(x, a=1, b=2, c=5)
```

```python
plt.figure(figsize=(12, 6))
```

```python
plt.plot(x, tri)
plt.title('Triangular')
plt.xlabel('x')
plt.ylabel('Membership Degree')
plt.legend()
plt.grid(True)
plt.ylim(-0.1, 1.1)
plt.show()
```

```python
plt.plot(x, trap)
plt.title('Trapezoidal')
plt.xlabel('x')
plt.ylabel('Membership Degree')
plt.legend()
plt.grid(True)
plt.ylim(-0.1, 1.1)
```

```python
plt.plot(x, bell_curve)
plt.title('Bell')
plt.xlabel('x')
plt.ylabel('Membership Degree')
plt.legend()
plt.grid(True)
plt.ylim(-0.1, 1.1)
```
