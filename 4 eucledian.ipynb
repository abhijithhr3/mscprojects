```python
import cv2
import numpy as np
```

```python
img1 = cv2.imread('/content/drive/MyDrive/download 1 .jpg', cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread('/content/drive/MyDrive/download 2.jpg', cv2.IMREAD_GRAYSCALE)
```

```python
img1 = cv2.resize(img1, (100, 100))
img2 = cv2.resize(img2, (100, 100))
```

```python
vec1 = img1.flatten()
vec2 = img2.flatten()
```

```python
img1
```

```python
img2
```

```python
distance = np.linalg.norm(vec1 - vec2)
print("Euclidean Distance between images:", distance)
```
