```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as LDA
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report
```

```python
train_df = pd.read_csv('/content/drive/MyDrive/MNIST/mnist_train.csv')
test_df = pd.read_csv('/content/drive/MyDrive/MNIST/mnist_test.csv')
```

```python
X_train = train_df.drop('label', axis=1).values
X_test = test_df.drop('label', axis=1).values
```

```python
y_train = train_df['label'].values
y_test = test_df['label'].values
```

```python
X_train = X_train / 255.0
X_test = X_test / 255.0
```

```python
pca = PCA(n_components=50)
X_train_pca = pca.fit_transform(X_train)
X_test_pca = pca.transform(X_test)
```

```python
lda = LDA(n_components=9)
X_train_lda = lda.fit_transform(X_train, y_train)
X_test_lda = lda.transform(X_test)
```

```python
svm_pca = SVC()
svm_pca.fit(X_train_pca, y_train)
y_pred_pca = svm_pca.predict(X_test_pca)
```

```python
svm_lda = SVC()
svm_lda.fit(X_train_lda, y_train)
y_pred_lda = svm_lda.predict(X_test_lda)
```

```python
print("PCA Accuracy:", accuracy_score(y_test, y_pred_pca))
print(classification_report(y_test, y_pred_pca))

print("LDA Accuracy:", accuracy_score(y_test, y_pred_lda))
print(classification_report(y_test, y_pred_lda))
```

```python
if(accuracy_score(y_test, y_pred_pca) > accuracy_score(y_test, y_pred_lda)):
  print("PCA has better Accuracy")
else:
  print("LDA has better Acccuracy")
```
