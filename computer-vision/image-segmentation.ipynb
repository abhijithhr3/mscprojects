```python
import matplotlib as mp
import matplotlib.pyplot as plt
import numpy as np
import cv2
from sklearn.cluster import KMeans
```

```python
img=np.zeros((520,650,3),np.uint8)
cv2.rectangle(img,(200,180),(550,400),(0,0,255),cv2.FILLED)
cv2.circle(img,(140,400),50,(0,255,0),cv2.FILLED)
cv2.rectangle(img,(100,150),(150,200),(255,0,0),cv2.FILLED)

```

```python
img.shape
```

```python
x=img.reshape(-1,3)
x.shape
```

```python
km=KMeans(n_clusters=3)
km.fit(x)
```

```python
segimg=km.cluster_centers_[km.labels_]
segimg=segimg.reshape(img.shape)
```

```python
plt.imshow(segimg/255)
```
