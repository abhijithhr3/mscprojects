```python
import tensorflow as tf
from tensorflow.keras.applications.resnet50 import ResNet50, preprocess_input, decode_predictions
from tensorflow.keras.preprocessing import image
import numpy as np
import matplotlib.pyplot as plt
```

```python
model = ResNet50(weights='imagenet')
```

```python
def predict_image(img_path):
    img = image.load_img(img_path, target_size=(224, 224))
    img_array = image.img_to_array(img)
    img_batch = np.expand_dims(img_array, axis=0)
    img_preprocessed = preprocess_input(img_batch)
    predictions = model.predict(img_preprocessed)
    label = decode_predictions(predictions, top=1)[0][0]
    plt.imshow(img)
    plt.axis("off")
    plt.title(f"Predicted: {label[1]}  (Confidence: {label[2]*100:.2f}%)")
    plt.show()
```

```python
predict_image("/content/drive/MyDrive/REsnet/Chihuahua.jpg")
predict_image("/content/drive/MyDrive/REsnet/baloon.jpg")
predict_image("/content/drive/MyDrive/REsnet/hen.jpg")
```
