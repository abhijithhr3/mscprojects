```python
import cv2
import numpy as np
```

```python
img1 = cv2.imread('/content/drive/MyDrive/iris image/iris setosa.jpg', cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread('/content/drive/MyDrive/iris image/iris versicolour.jpg', cv2.IMREAD_GRAYSCALE)

img1 = cv2.resize(img1, (256,256))
img2 = cv2.resize(img2, (256,256))
```

```python
flat1 = img1.flatten()
flat2 = img2.flatten()
```

```python
eucl_dist = 0
for i in range(len(flat1)):
    eucl_dist += (int(flat1[i]) - int(flat2[i])) ** 2

eucl_dist = eucl_dist ** 0.5
print("Euclidean Distance:", eucl_dist)
```

```python
mnht_dist = 0
for i in range(len(flat1)):
    mnht_dist += abs(int(flat1[i]) - int(flat2[i]))

print("Manhattan Distance:", mnht_dist)
```

```python
hmng_dist = 0
for i in range(len(flat1)):
    if flat1[i] != flat2[i]:
        hmng_dist += 1

print("Hamming Distance:", hmng_dist)
```
