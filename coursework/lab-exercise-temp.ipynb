```python
import numpy as np
!pip install -q scikit-fuzzy
import skfuzzy as fuzz
import matplotlib.pyplot as plt
```

```python
x_temp = np.arange(0, 71, 1)
```

```python
temp_low = fuzz.trimf(x_temp, [0, 20, 30])
temp_medium = fuzz.trimf(x_temp, [10, 30, 50])
temp_high = fuzz.trimf(x_temp, [30, 50, 70])
```

```python
input_temps = [10, 20, 30, 40, 50]
```

```python
print("Fuzzification Results (using skfuzzy):\n")
for t in input_temps:
    low = fuzz.interp_membership(x_temp, temp_low, t)
    medium = fuzz.interp_membership(x_temp, temp_medium, t)
    high = fuzz.interp_membership(x_temp, temp_high, t)

    print(f"Temperature {t}°C:")
    print(f"  → Low:    {round(low, 2)}")
    print(f"  → Medium: {round(medium, 2)}")
    print(f"  → High:   {round(high, 2)}\n")
```

```python
plt.figure(figsize=(12, 6))
```

```python
plt.plot(x_temp, temp_low, 'b', linewidth=1.5, label='Low')
plt.plot(x_temp, temp_medium, 'g', linewidth=1.5, label='Medium')
plt.plot(x_temp, temp_high, 'r', linewidth=1.5, label='High')
plt.title('Triangular')
plt.xlabel('x')
plt.ylabel('Membership Degree')
plt.legend()
plt.grid(True)
plt.ylim(-0.1, 1.1)
plt.show()
```
