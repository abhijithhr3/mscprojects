```python
A = dict()
B = dict()
union= dict()
inter= dict()
compl= dict()
```

```python
A = {"a": 0.2, "b": 0.3, "c": 0.6, "d": 0.6}
B = {"a": 0.9, "b": 0.9, "c": 0.4, "d": 0.5}

print('The First Fuzzy Set is :', A)
print('The Second Fuzzy Set is :', B)
```

```python
for key in A:
    union[key] = max(A[key], B[key])

print('Fuzzy Set Union is:', union)
```

```python
for key in A:
    inter[key] = min(A[key], B[key])

print('Fuzzy Set Intersection is:', inter)
```

```python
for key in A:
   compl[key]= 1-A[key]

print('Fuzzy Set Complement is :', compl)
```

```python
A_comp = {x: 1 - A[x] for x in A}
B_comp = {x: 1 - B[x] for x in B}
```

```python
# (A ∪ B)
left_1 = {x: 1 - union[x] for x in union}
#¬A ∩ ¬B
right_1 = {x: min(A_comp[x], B_comp[x]) for x in A}

# ¬(A ∩ B)
left_2 = {x: 1 - inter[x] for x in inter}
# ¬A ∪ ¬B
right_2 = {x: max(A_comp[x], B_comp[x]) for x in A}


print("A:", A)
print("B:", B)
print("\nDe Morgan's Law 1: ¬(A ∪ B) == ¬A ∩ ¬B")
print("LHS:", left_1)
print("RHS:", right_1)

print("\nDe Morgan's Law 2: ¬(A ∩ B) == ¬A ∪ ¬B")
print("LHS:", left_2)
print("RHS:", right_2)
```
