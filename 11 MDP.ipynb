```python
import numpy as np
```

```python
states = ['Rainy', 'Cloudy', 'Sunny']
actions = ['Umbrella', 'No Umbrella']
```

```python
P = {
    0: {  # Rainy
        0: [(0.6, 0, -1), (0.3, 1, -1), (0.1, 2, -1)],  # Umbrella
        1: [(0.6, 0, -10), (0.3, 1, -10), (0.1, 2, -10)]  # No Umbrella (gets wet!)
    },
    1: {  # Cloudy
        0: [(0.3, 0, -1), (0.4, 1, -1), (0.3, 2, -1)],  # Umbrella (inconvenient)
        1: [(0.3, 0, -5), (0.4, 1, 0), (0.3, 2, 0)]     # No Umbrella (risky if turns rainy)
    },
    2: {  # Sunny
        0: [(0.1, 0, -1), (0.2, 1, -1), (0.7, 2, -1)],  # Umbrella (unnecessary)
        1: [(0.1, 0, 0), (0.2, 1, 0), (0.7, 2, 0)]      # No Umbrella (ideal)
    }
}
```

```python
policy = {s: {a: 0.5 for a in actions} for s in states}
```

```python
V = {s: 0.0 for s in states}
```

```python
gamma = 0.9
```

```python
def policy_evaluation(V, P, policy, gamma=0.9, theta=1e-4):
    while True:
        delta = 0
        for s in states:
            v = V[s]
            new_v = 0
            for a in actions:
                prob_a = policy[s][a]
                for prob, next_state, reward in P[s][a]:
                    new_v += prob_a * prob * (reward + gamma * V[next_state])
            delta = max(delta, abs(v - new_v))
            V[s] = new_v
        if delta < theta:
            break
    return V
```

```python
V_result = policy_evaluation(V, P, policy, gamma)
```

```python
print("State Value Function under Random Policy:\n")
for s in states:
    print(f"{states[s]} (State {s}): {V_result[s]:.4f}")
```
